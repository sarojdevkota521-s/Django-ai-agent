```python
from ai.agent import get_movie_discovery_agent
```

```python
from langgraph.checkpoint.memory import InMemorySaver
checkpointer= InMemorySaver()
```

```python
agent =get_movie_discovery_agent(checkpointer=checkpointer)
```

```python
import uuid
config={"configurable":{"user_id":"4", "thread_id":uuid.uuid4()}}
response= agent.invoke({"messages":[{"role":"user","content":"Tell me about the first lord of the rings movie."}]},config)
```

```python

for i,msg in enumerate(response['messages']):
    if msg.content:
        print(i, msg.content, msg)
        print("/n/n")
```
