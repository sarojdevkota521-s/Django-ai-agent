```python
from ai.supervisors import get_supervisor
```

```python
from langgraph.checkpoint.memory import InMemorySaver
checkpointer= InMemorySaver()
```

```python
supervisor =get_supervisor(model= "gemini-3.7-flash", checkpointer=checkpointer)
```

```python
import uuid
config={"configurable":{"user_id":"4", "thread_id":uuid.uuid4()}}
response= supervisor.invoke({"messages":[{"role":"user","content":"What are my recent documents ?"}]},config)
```

```python

for i,msg in enumerate(response['messages']):
    if msg.content:
        print(i, msg.content, msg)
        print("/n/n")
```
